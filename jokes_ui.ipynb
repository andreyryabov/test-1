```python
from zenera import Agent, ModelSettings, create_python_mcp, init_project

## !

# Model assignments
model_router = "openai/responses/gpt-5-mini"
model_smart  = "openai/responses/gpt-5"

# Create MCP servers
py_mcp = await create_python_mcp()

# Specialist: UI Widgets (NiceGUI)
widgets = Agent(
    name="UI Widgets",
    instructions="widgets.md",
    model=model_smart,
    mcp_servers=[py_mcp],
    model_settings=ModelSettings(tool_choice="required"),
)

# Main agent: tells jokes, routes UI requests to Widgets
main = Agent(
    name="Main",
    instructions="main.md",
    model=model_router,
    handoffs=[widgets],
    model_settings=ModelSettings(tool_choice="auto"),
)

# Initialize project
await init_project(
    name="jokes-ui-assistant",
    description="Main agent that tells jokes and a UI Widgets agent that generates NiceGUI components.",
    default_agent=main,
    continuous_sync=True,
)
```
